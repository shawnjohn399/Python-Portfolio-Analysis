```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Stock": ["XOM", "AAPL", "MSFT"],
    "Price": [178.98, 343.46, 529.05],
    "Shares": [100, 50, 25]
}

portfolio = pd.DataFrame(data)

print(portfolio)
```

```python
portfolio["Market Value"] = portfolio["Price"] * portfolio["Shares"]
print(portfolio)
```

```python
total_value = portfolio["Market Value"].sum()
print("Total Portfolio Value: $", total_value)
```

```python
portfolio["Weight"] = portfolio["Market Value"] / total_value

print(portfolio)
```

```python
# Example purchase and current prices
purchase_prices = np.array([40, 300, 450])
current_prices = np.array([178.98, 343.46, 529.05])

# Calculate returns
returns = (current_prices - purchase_prices) / purchase_prices

print("Stock Returns:", returns)
print("Percentage Returns:", returns * 100)
```

```python
# Stock Returns:[3.4745 0.14486667 0.17566667]
# Percentage Returns:[347.45 14.48666667 17.56666667]
```

```python
stocks = ["XOM", "AAPL", "MSFT"]

for stock, ret in zip(stocks, returns):
    print(f"{stock}: {ret:.2%}")
```

```python
# Add the purchase prices to the portfolio DataFrame
portfolio["Purchase Price"] = purchase_prices

portfolio["Original Value"] = (
    portfolio["Purchase Price"] * portfolio["Shares"]
)

# Calculate profit for each stock
portfolio["Profit"] = (
    portfolio["Market Value"] - portfolio["Original Value"]
)

# Calculate overall portfolio return
total_invested = portfolio["Original Value"].sum()
total_profit = portfolio["Profit"].sum()

portfolio_return = total_profit / total_invested

print(portfolio[["Stock", "Original Value", "Market Value", "Profit"]])
print(f"Total Invested: ${total_invested:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Portfolio Return: {portfolio_return:.2%}")
```
